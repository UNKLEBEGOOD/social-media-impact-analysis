```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
df = pd.read_csv('data/clean/social_media_impact_clean.csv')
```

**usage vs mental health (your strongest correlation, -0.85)**

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Daily_Usage_Hours', y='Mental_Health_Index', alpha=0.4, color='#2c7fb8')
plt.title('Higher Social Media Usage Is Linked to Lower Mental Wellbeing', fontsize=13)
plt.xlabel('Daily Usage (Hours)')
plt.ylabel('Mental Health Index')
plt.tight_layout()
plt.savefig('visuals/mental_health_vs_usage.png', dpi=200)
plt.show()
```

**SOCIAL MEDIA USAGE IS LINKED TO POOR SLEEEP**

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Daily_Usage_Hours', y='Sleep_Duration_Hours', alpha=0.4, color='#2c7fb8')
plt.title('Higher Social Media Usage Is Linked to Poor Sleep', fontsize=13)
plt.xlabel('Daily Usage (Hours)')
plt.ylabel('Sleep Duration (Hours)')
plt.tight_layout()
plt.savefig('visuals/sleep_duration_vs_usage.png', dpi=200)
plt.show()
```

**HIGHER SOCIAL MEDIA USAGE VS LOWER GP**

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Daily_Usage_Hours', y='Academic_Performance_GPA', alpha=0.4, color='#2c7fb8')
plt.title('Higher Social Media Usage Is Linked to Lower GPA', fontsize=13)
plt.xlabel('Daily Usage (Hours)')
plt.ylabel('GPA')
plt.tight_layout()
plt.savefig('visuals/gpa_vs_usage.png', dpi=200)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Daily_Usage_Hours', y='Perceived_Stress_Score', alpha=0.4, color='#de2d26')
plt.title('Higher Social Media Usage Is Linked to Higher Stress', fontsize=13)
plt.xlabel('Daily Usage (Hours)')
plt.ylabel('Perceived Stress Score')
plt.tight_layout()
plt.savefig('visuals/stress_vs_usage.png', dpi=200)
plt.show()
```

**Overall Impact box plot (your headline chart):**

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Overall_Impact', y='Daily_Usage_Hours', hue='Overall_Impact', 
            order=['Beneficial', 'Neutral', 'Negative'], 
            palette=['#2ca25f', '#fec44f', '#de2d26'], legend=False)
plt.title('Students Reporting Negative Impact Use Social Media Nearly 3x More', fontsize=13)
plt.xlabel('Self-Reported Overall Impact')
plt.ylabel('Daily Usage (Hours)')
plt.tight_layout()
plt.savefig('visuals/impact_vs_usage.png', dpi=200)
plt.show()
```

**DAILY USAGE BY PRIMARY PLATFORM**

```python
plt.figure(figsize=(9,5))
platform_order = df.groupby('Primary_Platform')['Daily_Usage_Hours'].mean().sort_values(ascending=False).index

sns.boxplot(data=df, x='Primary_Platform', y='Daily_Usage_Hours', 
            hue='Primary_Platform', order=platform_order, legend=False,
            palette='Set2')
plt.title('Daily Usage by Primary Platform')
plt.xlabel('Primary Platform')
plt.ylabel('Daily Usage (Hours)')
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig('visuals/platform_vs_usage.png', dpi=200)
plt.show()
```
