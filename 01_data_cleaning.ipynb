## Outlier Check
Checked Age, Daily_Usage_Hours, Weekend_Extra_Hours, Sleep_Duration_Hours, 
Perceived_Stress_Score, and Academic_Performance_GPA using the IQR method. 
All flagged values fall within realistic human ranges (no negative hours, 
no impossible ages, no GPA above 4.0) — so no values were removed or capped. 
These are treated as genuine variation, not data errors.

```python
def check_outliers(col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col}: bounds ({lower:.2f}, {upper:.2f}) → {len(outliers)} outliers")

for col in ['Age', 'Daily_Usage_Hours', 'Weekend_Extra_Hours', 'Sleep_Duration_Hours', 'Perceived_Stress_Score', 'Academic_Performance_GPA']:
    check_outliers(col)
```

## Missing Value Treatment
- `Perceived_Stress_Score` (46 missing) and `Academic_Performance_GPA` (85 missing) 
  were filled using the median value within each student's Academic_Level group, 
  rather than dropping rows or using a single overall average. This preserves 
  differences between High School, Undergraduate, and Postgraduate students 
  while avoiding data loss.
- No duplicate rows were found.
- Outlier check (IQR method) found no impossible values across numeric columns — 
  all flagged values were retained as genuine variation.

```python
df['Perceived_Stress_Score'] = df.groupby('Academic_Level')['Perceived_Stress_Score'].transform(
    lambda x: x.fillna(x.median())
)

df['Academic_Performance_GPA'] = df.groupby('Academic_Level')['Academic_Performance_GPA'].transform(
    lambda x: x.fillna(x.median())
)
```
